```python
import cv2, tqdm
import pandas as pd
```

```python
df = pd.read_csv('wiki_imdb_lfw.csv')
```

```python
H, W, C = [], [], []
for i, path in tqdm.tqdm(enumerate(df.path)):
    try:
        img = cv2.imread(path)
        if img is not None:
            height, width, channel = img.shape
        else:
            height, width, channel = None, None, None
        H.append(height)
        W.append(width)
        C.append(channel)
    except FileNotFoundError as e:
        print(f"{i} Error: {e}")
```

```python
df['height'] = H
df['width'] = W
df['channels'] = C
```

# Discard not appropriate images from dataset

Policy: 
1. Image is not found;
2. Image height or width is less than 32 pixels.

```python
include = []

for i in tqdm.trange(0, len(df)):
    height = df.height[i]
    width = df.width[i]
    if height is None:
        include.append(False)
        continue
    elif height < 32 or width < 32:
        include.append(False)
        continue
    include.append(True)
```

```python
df['include'] = include
# df.to_csv('wiki_imdb_lfw.csv', index=False)
```
